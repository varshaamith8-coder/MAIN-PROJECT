<h2 style="background-color: lightblue; padding: 10px;">
Column Description
</h2>

```python
VIN (1-10):
The first 10 characters of the vehicle identification number used to identify the vehicle

County:
The county where the electric vehicle is registered

City:
The city where the electric vehicle is registered.

State:
The state where the electric vehicle is registered.

Postal Code:
The postal/ZIP code associated with the vehicle's registration location.

Model Year:
The year in which the electric vehicle model was manufactured.

Make:
The manufacturer or brand of the electric vehicle.

Model:
The specific model name of the electric vehicle.

Electric Vehicle Type:
Indicates whether the vehicle is a Battery Electric Vehicle (BEV) or Plug-in Hybrid Electric Vehicle (PHEV).

Clean Alternative Fuel Vehicle (CAFV) Eligibility:
Indicates whether the vehicle meets the eligibility requirements for clean alternative fuel vehicle programs.

Electric Range:
The estimated distance the vehicle can travel using electric power on a full charge, measured in miles.

Base MSRP:
The manufacturer's suggested retail price of the vehicle's base model.

Legislative District:
The political legislative district in which the vehicle is registered.

DOL Vehicle ID:
The unique identification number assigned to the vehicle by the Department of Licensing.

Vehicle Location:
The geographic location of the vehicle registration, generally represented using latitude and longitude coordinates.

Electric Utility:
The electric utility company serving the vehicle's registered location.

2020 Census Tract:
The census tract in which the vehicle is registered, based on 2020 Census geographic boundaries.
```

# Data Pre-Processing

```python
# Handel missing values
# For numerivcal columns use median
numerical_columns = [
    'Postal Code',
    'Electric Range',
    'Legislative District',
    '2020 GEOID']
for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())
```

```python
# For Categorical columns use mode
categorical_columns = [
    'County',
    'City',
    'Vehicle Location',
    'Electric Utility']
for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])
```

```python
df.isnull().sum()
```

## No duplicates rows

```python
# correct data types

df.dtypes
```

```python
df['Postal Code'] = df['Postal Code'].astype(int)
df['Model Year'] = df['Model Year'].astype(int)
df['Electric Range'] = df['Electric Range'].astype(float)
df['Legislative District'] = df['Legislative District'].astype(int)
df['DOL Vehicle ID'] = df['DOL Vehicle ID'].astype(int)
df['2020 GEOID'] = df['2020 GEOID'].astype(int)
```

```python
df.dtypes
```

```python
# creating derived columns
# Create vehicle age

current_year = 2026
df['Vehicle Age'] = current_year - df['Model Year']
```

```python
df[['Model Year', 'Vehicle Age']].head()
```

```python
# Create range category
def range_category(x):
    if x < 100:
        return 'Low Range'
    elif x < 200:
        return 'Medium Range'
    else:
        return 'High Range'
df['Range Category'] = df['Electric Range'].apply(range_category)
```
