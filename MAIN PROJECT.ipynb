Column Description
</h2>

```python
VIN (1-10):
The first 10 characters of the vehicle identification number used to identify the vehicle

County:
The county where the electric vehicle is registered

City:
The city where the electric vehicle is registered.

State:
The state where the electric vehicle is registered.

Postal Code:
The postal/ZIP code associated with the vehicle's registration location.

Model Year:
The year in which the electric vehicle model was manufactured.

Make:
The manufacturer or brand of the electric vehicle.

Model:
The specific model name of the electric vehicle.

Electric Vehicle Type:
Indicates whether the vehicle is a Battery Electric Vehicle (BEV) or Plug-in Hybrid Electric Vehicle (PHEV).

Clean Alternative Fuel Vehicle (CAFV) Eligibility:
Indicates whether the vehicle meets the eligibility requirements for clean alternative fuel vehicle programs.

Electric Range:
The estimated distance the vehicle can travel using electric power on a full charge, measured in miles.

Base MSRP:
The manufacturer's suggested retail price of the vehicle's base model.

Legislative District:
The political legislative district in which the vehicle is registered.

DOL Vehicle ID:
The unique identification number assigned to the vehicle by the Department of Licensing.

Vehicle Location:
The geographic location of the vehicle registration, generally represented using latitude and longitude coordinates.

Electric Utility:
The electric utility company serving the vehicle's registered location.

2020 Census Tract:
The census tract in which the vehicle is registered, based on 2020 Census geographic boundaries.
```

```python
# Handel missing values
# For numerivcal columns use median
numerical_columns = [
    'Postal Code',
    'Electric Range',
    'Legislative District',
    '2020 GEOID']
for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())
```

## Outlier Checking

```python
numeric_columns = df.select_dtypes(include=np.number).columns
numeric_columns
```

```python
for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    outliers = df[(df[column] < lower_limit) | (df[column] > upper_limit)]
    
    print(f"\n{column}")
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
    print("Number of outliers:", len(outliers))
```

```python
# Summary
outlier_summary = []

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    count = ((df[column] < lower_limit) | 
             (df[column] > upper_limit)).sum()
    
    outlier_summary.append([
        column, Q1, Q3, IQR, lower_limit, upper_limit, count
    ])

outlier_df = pd.DataFrame(
    outlier_summary,
    columns=[
        'Column', 'Q1', 'Q3', 'IQR',
        'Lower Limit', 'Upper Limit', 'Outlier Count'
    ]
)

outlier_df
```

### Pivot table analysis

```python
# Vehicle type across state
top_states = df['State'].value_counts().head(5).index
pivot = pd.pivot_table(
    df[df['State'].isin(top_states)],
    index='State',
    columns='Electric Vehicle Type',
    values='DOL Vehicle ID',
    aggfunc='count',
    fill_value=0)
pivot
```

```python
pivot.plot(
    kind='bar',
    figsize=(10,6))
plt.title('EV Type Distribution Across Top States')
plt.xlabel('State')
plt.ylabel('Number of Vehicles')
plt.xticks(rotation=0)
plt.legend(title='Vehicle Type')
plt.show()
```

#### Insight

BEVs outnumber PHEVs in each of the major states represented in the dataset. Washington has by far the largest number of both vehicle types because it contains almost all records in the dataset.
